# Fake News Detection (TF-IDF + SGD)
Run cells in order. Expects `../data/news.csv` with `text` and `label` columns.

```
import sys
sys.path.append('../src')
from train_model import load_data, build_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

```
df = load_data('../data/news.csv')
df['label'].value_counts()
```

```
x_train, x_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.2, random_state=42, stratify=df['label'])
pipe = build_pipeline().fit(x_train, y_train)
y_pred = pipe.predict(x_test)
print(f'Accuracy: {accuracy_score(y_test, y_pred):.2%}')
print(classification_report(y_test, y_pred))
```

```
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, labels=['FAKE','REAL']);
```
